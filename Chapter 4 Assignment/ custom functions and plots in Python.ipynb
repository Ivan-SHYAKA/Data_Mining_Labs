```python
import numpy as np
import matplotlib.pyplot as plt
```

### A) Fixed Function

```python
def Power():
    result = 2 ** 3
    print(result)
Power()
```

### B) Dynamic Printing Function

```python
def Power2(x, a):
    result = x ** a
    print(result)
Power2(3, 8)
```

### C) Calculations Using Power2

```python
print("10^3 = ", end ="")
Power2(10, 3)
print("8^17 = ", end="")
Power2(8, 17)
print("13^13 = ", end="")
Power2(13, 13)
```

### D) Object Returning Function

```python
def Power3(x, a):
    result = x ** a
    return result
```

### E) Plotting x^2 with Subplots

```python
x_vals_e = np.arange(1, 11)
# feeding values to our Power3 function
y_vals_e = [Power3(x, 2) for x in x_vals_e]

# side-by-side plot layout
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1
axes[0].plot(x_vals_e, y_vals_e, marker='o', color='b')
axes[0].set_xlabel('Integers (x)')
axes[0].set_ylabel('x squared (y)')
axes[0].set_title('Linear Scale Plot of f(x) = x^2')
axes[0].grid(True)

# Plot 2
axes[1].plot(x_vals_e, y_vals_e, marker='s', color='r')
axes[1].set_xscale('log')
axes[1].set_yscale('log')
axes[1].set_xlabel('Log of x')
axes[1].set_ylabel('Log of x squared')
axes[1].set_title('Log-Log Scale Plot of f(x) = x^2')
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

### F) Generalized Plotting Function

```python
def PlotPower(x_range, exponent):
    y_range = [Power3(x, exponent) for x in x_range]
    
    plt.figure(figsize=(7, 4.5))
    plt.plot(x_range, y_range, marker='o', linestyle='--', color='purple')
    plt.xlabel('x sequence values')
    plt.ylabel(f'x raised to the power of {exponent}')
    plt.title(f'Custom Power Plot: f(x) = x^{exponent}')
    plt.grid(True)
    plt.show()

# Calling the function as requested in the book 
PlotPower(np.arange(1, 11), 3)
```
